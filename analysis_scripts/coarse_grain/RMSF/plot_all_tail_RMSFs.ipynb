```python
tab20 = plt.get_cmap("tab20")

# Sequences (one-letter)
H3_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKPHRYRPGTVALREIRRY")
H3withcenpahelix_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKPHRYRPGQGWLKEIRKL")
H3withcenpalatch_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKHQHSRRRTVALREIRRY")
stcenpacore_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKHQHSRRRQGWLKEIRKL")
H3withcenpa_latch_helix_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKHQHSRRRQGWLKEIRKL")
ltcenpacore_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKPHRYRPGQGWLKEIRKL")
thruhelixcenpa_seq = list("ARTKQTARKSTGGKAPRKQLATKAARKSAPATGGVKKPHRYRPGTVALREIRRY")

nres = len(H3_seq)  # Tail residues
x = np.arange(nres)

# Load RMSF data (columns: RMSF and error)
y_H3_all, yerr_H3_all = np.loadtxt("H3/H3tail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_H3withcenpahelix_all, yerr_H3withcenpahelix_all = np.loadtxt("H3withcenpahelix/H3withcenpahelixtail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_H3withcenpalatch_all, yerr_H3withcenpalatch_all = np.loadtxt("H3withcenpalatch/H3withcenpalatchtail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_st_all, yerr_st_all = np.loadtxt("stcenpacore/stcenpacoretail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_H3withcenpa_latch_helix_all, yerr_H3withcenpa_latch_helix_all = np.loadtxt("H3withcenpa_latch_helix/H3withcenpa_latch_helixtail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_lt_all, yerr_lt_all = np.loadtxt("ltcenpacore/ltcenpacoretail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))
y_thruhelixcenpa_all, yerr_thruhelixcenpa_all = np.loadtxt("thruhelixcenpa/thruhelixcenpatail.dat", usecols=(1,2), unpack=True, comments=('#','@','&'))

# Correct split for H3 tails: exit 1-54, entry 55-108
y_H3_entry = y_H3_all[54:108]
yerr_H3_entry = yerr_H3_all[54:108]

y_H3withcenpahelix_entry = y_H3withcenpahelix_all[54:108]
yerr_H3withcenpahelix_entry = yerr_H3withcenpahelix_all[54:108]

y_H3withcenpalatch_entry = y_H3withcenpalatch_all[54:108]
yerr_H3withcenpalatch_entry = yerr_H3withcenpalatch_all[54:108]

y_st_entry = y_st_all[54:108]
yerr_st_entry = yerr_st_all[54:108]

y_H3withcenpa_latch_helix_entry = y_H3withcenpa_latch_helix_all[54:108]
yerr_H3withcenpa_latch_helix_entry = yerr_H3withcenpa_latch_helix_all[54:108]

y_lt_entry = y_lt_all[54:108]
yerr_lt_entry = yerr_lt_all[54:108]

y_thruhelixcenpa_entry = y_thruhelixcenpa_all[54:108]
yerr_thruhelixcenpa_entry = yerr_thruhelixcenpa_all[54:108]


# Figure
fig, ax = plt.subplots(figsize=(12,9))

# Plot entry tails (opaque)
line_H3_entry, = ax.plot(x, y_H3_entry, color=tab20(0), lw=2)
line_H3withcenpahelix_entry, = ax.plot(x, y_H3withcenpahelix_entry, color=tab20(18), lw=2)
line_H3withcenpalatch_entry, = ax.plot(x, y_H3withcenpalatch_entry, color=tab20(1), lw=2)
line_st_entry, = ax.plot(x, y_st_entry, color=tab20(7), lw=2)
line_H3withcenpa_latch_helix_entry, = ax.plot(x, y_H3withcenpa_latch_helix_entry, color=tab20(13), lw=2)
line_lt_entry, = ax.plot(x, y_lt_entry, color=tab20(6), lw=2)
line_thruhelixcenpa_entry, = ax.plot(x, y_thruhelixcenpa_entry, color=tab20(12), lw=2)

# Shaded error
def fill_err(x, y, yerr, color, alpha):
    ax.fill_between(x, y-yerr/2, y+yerr/2, color=color, alpha=alpha)

fill_err(x, y_H3_entry, yerr_H3_entry, tab20(0), 0.3)
fill_err(x, y_H3withcenpahelix_entry, yerr_H3withcenpahelix_entry, tab20(18), 0.3)
fill_err(x, y_H3withcenpalatch_entry, yerr_H3withcenpalatch_entry, tab20(1), 0.3)
fill_err(x, y_st_entry, yerr_st_entry, tab20(7), 0.3)
fill_err(x, y_H3withcenpa_latch_helix_entry, yerr_H3withcenpa_latch_helix_entry, tab20(13), 0.3)
fill_err(x, y_lt_entry, yerr_lt_entry, tab20(6), 0.3)
fill_err(x, y_thruhelixcenpa_entry, yerr_thruhelixcenpa_entry, tab20(12), 0.3)

bottom_labels = []
for i in range(54):
    if i < 37:
        bottom_labels.append(f"{H3_seq[i]}{i+1}")
    else:
        bottom_labels.append(f"{H3_seq[i]}{i+1}")

ax.set_xticks(x)
ax.set_xticklabels(bottom_labels, rotation=90, fontsize=8)

# Color H3 labels
for i, tick in enumerate(ax.get_xticklabels()):
    if i < 37:
        tick.set_color('black')
    else:
        tick.set_color(tab20(0))

# Add other sequences below for residues 45-54
for i in range(37,54):
    ax.text(i, -6.25, f"{H3withcenpahelix_seq[i]}{i+1}", color=tab20(18), ha='center', va='top', rotation=90, fontsize=8)
    ax.text(i, -12.50, f"{H3withcenpalatch_seq[i]}{i+1}", color=tab20(1), ha='center', va='top', rotation=90, fontsize=8)
    ax.text(i, -18.75, f"{stcenpacore_seq[i]}{i+1}", color=tab20(7), ha='center', va='top', rotation=90, fontsize=8)
    ax.text(i, -25.00, f"{H3withcenpa_latch_helix_seq[i]}{i+1}", color=tab20(13), ha='center', va='top', rotation=90, fontsize=8)
    ax.text(i, -31.25, f"{ltcenpacore_seq[i]}{i+1}", color=tab20(6), ha='center', va='top', rotation=90, fontsize=8)
    ax.text(i, -37.50, f"{thruhelixcenpa_seq[i]}{i+1}", color=tab20(12), ha='center', va='top', rotation=90, fontsize=8)

# Tail/Latch/Helix divider
ax.axvline(36.5, color='k', linestyle='--', lw=1)
ax.axvline(43.5, color='k', linestyle='--', lw=1)

# Formatting
ax.set_ylabel(r"RMSF ($\mathrm{\AA}$)", fontsize=12)
ax.set_yticks(np.arange(0, 50, 8))
ax.set_yticklabels(np.arange(0, 50, 8), fontsize=10)

ax.set_xlim(0,53)
ax.set_ylim(0,48)

# Legend (entry tails first)
ax.legend([line_H3_entry, line_H3withcenpahelix_entry, line_H3withcenpalatch_entry, line_st_entry,
           line_H3withcenpa_latch_helix_entry, line_lt_entry, line_thruhelixcenpa_entry],
          ['H3 entry',
           r'H3 with CENP-A$_{\mathrm{\alpha N}}$ entry',
           r'H3 with CENP-A$_{\mathrm{Latch}}$ entry',
           r'H3$_{\mathrm{Tail}}$CENP-A$_{\mathrm{Core}}$ entry',
           r'CENP-A$_{\mathrm{Latch+\alpha N}}$H3$_{\mathrm{Core}}$ entry',
           r'H3$_{\mathrm{Tail+Latch}}$CENP-A$_{\mathrm{Core}}$ entry',
           r'H3$_{\mathrm{Tail+Latch+\alpha N}}$CENP-A$_{\mathrm{Core}}$ entry'],
          fontsize=10, loc='upper right')

plt.title('H3 CG Entry Tails RMSF', fontsize=18, fontweight='bold')
plt.tight_layout()
plt.savefig('CG_H3_tails_RMSF_entry.pdf')
plt.show()
```
